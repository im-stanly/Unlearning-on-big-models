```python
import matplotlib.pyplot as plt
%matplotlib inline
```

### Input data for concept removal plot

```python
from src.utils import load_images

ship_images = load_images("../assets/ship")
ship_image_masks = load_images("../assets/ship_mask")

n_cols = len(ship_images)
n_rows = 2

fig, ax = plt.subplots(nrows=n_rows, ncols=n_cols, figsize=(16, 8), sharex=True, sharey=True)

fig.suptitle("Training data: concepts for removal with their respective masks")

for n, (img, mask) in enumerate(zip(ship_images, ship_image_masks)):
    ax[0, n].imshow(img)
    ax[1, n].imshow(mask, cmap="gray")
    ax[0, n].set(xticks=[], yticks=[])
    ax[1, n].set(xticks=[], yticks=[])

plt.subplots_adjust(wspace=0, hspace=0)
fig.tight_layout()
plt.savefig("../assets/ships_with_masks_plot.png")
plt.show()
```
